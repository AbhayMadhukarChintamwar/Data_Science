```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('fligh_price.csv')
print(df.head())

df['Date_Of_Journey'].str.split('-')
df['Date']= df['Date_Of_Journey'].str.split('-').str[2]
df['Month'] = df['Date_Of_Journey'].str.split('-').str[1]
df['Year']= df['Date_Of_Journey'].str.split('-').str[0]

df['Departure_Hours']= df['Departure_Time'].str.split(' ').str[1].str.split(':').str[0]
df['Departure_Minutes']=df['Departure_Time'].str.split(' ').str[1].str.split(':').str[1]
df['Departure_Seconds']=df['Departure_Time'].str.split(' ').str[1].str.split(':').str[2]

df['Arrivala_Hours'] = df['Arrivala_Time'].str.split(' ').str[1].str.split(':').str[0]
df['Arrivala_Minutes'] = df['Arrivala_Time'].str.split(' ').str[1].str.split(':').str[1]
df['Arrivala_Seconds'] = df['Arrivala_Time'].str.split(' ').str[1].str.split(':').str[2]


print(df.head())
print(df.info())

df['Date'] =df['Date'].astype(int)
df['Month'] = df['Month'].astype(int)
df['Year'] = df['Year'].astype(int)
df['Departure_Hours'] = df['Departure_Hours'].astype(int)
df['Departure_Minutes'] =df['Departure_Minutes'].astype(int)
df['Departure_Seconds'] = df['Departure_Seconds'].astype(int)
df['Arrivala_Hours'] =df['Arrivala_Hours'].astype(int)
df['Arrivala_Minutes'] =df['Arrivala_Minutes'].astype(int)
df['Arrivala_Seconds'] = df['Arrivala_Seconds'].astype(int)
print(df.head())

print(df.info())

df.drop('Date_Of_Journey', axis=1, inplace=True)
df.drop('Route', axis=1, inplace=True)
df.drop('Departure_Time', axis=1, inplace = True)
df.drop('Arrivala_Time', axis = 1, inplace = True)
df.drop('Duration', axis = 1, inplace = True)
df.drop('Days_Left', axis  = 1, inplace = True)

print(df['Total_Stops'].unique())


df['Total_Stops'] = df['Total_Stops'].map({
    'Non-stop':0,
    '1 Stop' :1,
    '2+ Stops' :2  
})

print(df.head())



print(df['Airline'].unique())

print(df['Source_City'].unique())

print(df['Destination'].unique())


print(pd.DataFrame(df.head()))

df.head()

from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder()
encoder.fit_transform(df[['Airline','Source_City','Destination']]).toarray()

data = pd.DataFrame(encoder.fit_transform(df[['Airline','Source_City','Destination']]).toarray(),columns = encoder.get_feature_names_out())

print(data)
print('<----------------------------------------------------------------------------------------------------------------------------->')
```
