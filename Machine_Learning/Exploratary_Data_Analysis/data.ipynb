```python
# Install if not installed
# pip install pandas numpy

import pandas as pd
import numpy as np

# Number of records
num_rows = 1000

# Categories
gender = ['Male', 'Female']

race_ethnicity = [
    'Group A',
    'Group B',
    'Group C',
    'Group D',
    'Group E'
]

parental_education = [
    "bachelor's degree",
    'some college',
    "master's degree",
    "associate's degree",
    'high school'
]

lunch = [
    'standard',
    'free/reduced'
]

test_preparation_course = [
    'completed',
    'none'
]

# Generate random data
data = {
    'gender': np.random.choice(gender, num_rows),

    'race/ethnicity': np.random.choice(
        race_ethnicity, num_rows
    ),

    'parental level of education': np.random.choice(
        parental_education, num_rows
    ),

    'lunch': np.random.choice(
        lunch, num_rows
    ),

    'test preparation course': np.random.choice(
        test_preparation_course, num_rows
    ),

    'math score': np.random.randint(
        0, 101, num_rows
    ),

    'reading score': np.random.randint(
        0, 101, num_rows
    ),

    'writing score': np.random.randint(
        0, 101, num_rows
    )
}

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
df.to_csv('stud.csv', index=False)

# Show first 5 rows
print("stud.csv file created successfully!")
print(df)
```
