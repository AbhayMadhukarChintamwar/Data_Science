```python
# Install pandas if not installed
# pip install pandas numpy

import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Number of rows
num_rows = 1000

# Airlines (6 airlines)
airlines = [
    "IndiGo",
    "Air India",
    "SpiceJet",
    "Vistara",
    "Go First",
    "Akasa Air"
]

# Flight codes
flight_prefix = {
    "IndiGo": "6E",
    "Air India": "AI",
    "SpiceJet": "SG",
    "Vistara": "UK",
    "Go First": "G8",
    "Akasa Air": "QP"
}

# Cities (6 cities)
cities = [
    "Delhi",
    "Mumbai",
    "Bangalore",
    "Kolkata",
    "Hyderabad",
    "Chennai"
]

# Stops
stops_options = ["Non-stop", "1 Stop", "2+ Stops"]

# Seat class
seat_class = ["Economy", "Business"]

# Data list
data = []

# Base booking date
booking_start = datetime(2025, 1, 1)

for _ in range(num_rows):

    # Airline
    airline = random.choice(airlines)

    # Flight code
    flight = f"{flight_prefix[airline]}{random.randint(100, 9999)}"

    # Source and destination
    source = random.choice(cities)
    destination = random.choice([city for city in cities if city != source])

    # Route
    route = f"{source} -> {destination}"

    # Journey date
    days_left = random.randint(1, 60)

    booking_date = booking_start + timedelta(days=random.randint(0, 180))
    journey_date = booking_date + timedelta(days=days_left)

    # Departure datetime (real date and time)
    dep_hour = random.randint(0, 23)
    dep_minute = random.choice([0, 10, 15, 20, 30, 45, 50])

    departure_time = journey_date.replace(
        hour=dep_hour,
        minute=dep_minute,
        second=0
    )

    # Stops
    total_stops = random.choice(stops_options)

    # Duration based on stops
    if total_stops == "Non-stop":
        duration_hours = random.randint(1, 4)
    elif total_stops == "1 Stop":
        duration_hours = random.randint(3, 8)
    else:
        duration_hours = random.randint(6, 12)

    duration_minutes = random.randint(0, 59)

    duration = timedelta(
        hours=duration_hours,
        minutes=duration_minutes
    )

    # Arrival datetime
    arrival_time = departure_time + duration

    # Seat class
    travel_class = random.choice(seat_class)

    # Price generation
    base_price = random.randint(2500, 8000)

    # Business class price increase
    if travel_class == "Business":
        base_price *= random.randint(2, 4)

    # Stops effect
    if total_stops == "1 Stop":
        base_price += random.randint(1000, 3000)
    elif total_stops == "2+ Stops":
        base_price += random.randint(3000, 6000)

    # Airline effect
    if airline == "Vistara":
        base_price += 2000
    elif airline == "Air India":
        base_price += 1500

    price = base_price

    # Save row
    data.append([
        airline,
        journey_date.strftime("%Y-%m-%d"),
        flight,
        source,
        destination,
        route,
        departure_time.strftime("%Y-%m-%d %H:%M:%S"),
        arrival_time.strftime("%Y-%m-%d %H:%M:%S"),
        total_stops,
        travel_class,
        f"{duration_hours}h {duration_minutes}m",
        days_left,
        price
    ])

# Create DataFrame
columns = [
    "Airline",
    "Date_Of_Journey",
    "Flight",
    "Source City",
    "Destination",
    "Route",
    "Departure Time",
    "Arrival Time",
    "Total_Stops",
    "Class",
    "Duration",
    "Days Left",
    "Price"
]

df = pd.DataFrame(data, columns=columns)

# Save CSV
df.to_csv("fligh_price.csv", index=False)

print("CSV file 'fligh_price.csv' created successfully!")
print(df.head())
```
