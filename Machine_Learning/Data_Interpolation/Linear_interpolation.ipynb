Linear Interpolation

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])

plt.scatter(x,y)
plt.show()

```

```python
x_new = np.linspace(1,5, 10)
x_new
```

```python
y_interp = np.interp(x_new, x, y)
y_interp
```

```python
plt.scatter(x_new, y_interp, c='r')
plt.plot(x,y)
plt.show()
```
