Cubic Interpolation

```python
import numpy as np
x= np.array([1,2,3,4,5])
y = np.array([1,8,27,64,125])
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
plt.show()
```

```python
from scipy.interpolate import interp1d
```

```python
f =interp1d(x, y, kind='cubic')
```

```python
x_new = np.linspace(1,5,20)
y_interp = f(x_new)
```

```python
y_interp
```

```python
plt.scatter(x_new, y_interp)
plt.scatter(x,y)
plt.show()
```
