Polynomial Data Interpolation

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1,2,3,4,5])
y = np.array([1,4,9,16,25])
plt.scatter(x,y)
plt.show()
```

```python
p =np.polyfit(x,y,2)
```

```python
x_new = np.linspace(1, 5, 20)
y_interp = np.polyval(p,x_new)
```

```python
plt.scatter(x_new, y_interp)
plt.show()
```

```python
plt.scatter(x_new, y_interp)
plt.scatter(x,y)
plt.show()
```

```python
plt.scatter(x_new, y_interp)
plt.scatter(x,y)
plt.grid()
plt.show()
```

```python
p =np.polyfit(x,y,4)
```

```python
x_new = np.linspace(1, 5, 20)
y_interp = np.polyval(p,x_new)
plt.scatter(x_new, y_interp)
plt.scatter(x,y)
plt.grid()
plt.show()
```
