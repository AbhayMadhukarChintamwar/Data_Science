Handling Outerlier

```python
import numpy as np
```

```python
lst_marks = [45,32,56,75,89,54,32,89,90,87,67,54,45,98,99,67,74,1000,1100]
```

```python
np.percentile(lst_marks, [25])
```

```python
np.percentile(lst_marks, [75])
```

```python
minimum , Q1, Q2, Q3, maximum = np.quantile(lst_marks,[0,0.25,0.50,0.75,1.0])
```

```python
minimum
```

```python
Q2
```

```python
maximum
```

```python
IQR = Q3-Q1
IQR
```

```python
lower_fench = Q1-1.5*(IQR)
higher_fench = Q3 + 1.5*(IQR)
```

```python
lower_fench
```

```python
higher_fench
```

```python
outliers = []
for i in lst_marks:
    if i>=lower_fench and i<=higher_fench:
        print('This element is not an outlier')
    else:
        outliers.append(i)
      
```
