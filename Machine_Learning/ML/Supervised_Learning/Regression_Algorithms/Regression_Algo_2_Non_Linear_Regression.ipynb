#### Linear Regression Algorithms

```python
import pandas as pd

df = pd.DataFrame({
    'inp' :[1,2,3,4,5],
    'out':[1,4,9,16,25]
})

x = df[['inp']]
y = df['out']

# Modelling (Model Selection)
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x, y)
print('Coefficien of Linear Regression Model:', model.coef_)
print('Intercept of Linear Regression Model', model.intercept_)

#  Prediction

y_pred = model.predict(x)

# Evaluation 
from sklearn.metrics import mean_squared_error
from math import sqrt
mse  = mean_squared_error(y, y_pred)
rmse = sqrt(mse)

print('Mean Square Error in Linear Regression :',mse)
print('Root Mean Square Error in Linear Regression :',rmse)
df
```

#### Non-Linear Regression Algorithms

```python
import pandas as pd

df = pd.DataFrame({
    'inp' :[1,2,3,4,5],
    'out':[1,4,9,16,25]
})
df['inp2'] = df['inp']**2

x = df[['inp2','inp']]
y = df['out']
df
# modelling

from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x,y)
print('Coefficient of Non Linear Regression :', model.coef_)
print('Intercept of Non Linear Regression :', model.intercept_)

# Prediction
y_pred = model.predict(x)

# Evaluation
from sklearn.metrics import mean_squared_error
from math import sqrt

mse = mean_squared_error(y, y_pred)
rmse = sqrt(mse)

print('Mean Squared Error in Non Linear Regression :', mse)
print('Root Mean Squared Erro in Non Linear Regression :', rmse)
```
