## 1. What is Hierarchical Clustering?

**Hierarchical Clustering** is an **unsupervised machine learning algorithm** that groups similar data points into clusters by creating a **hierarchical tree-like structure** called a **dendrogram**.

### Professional Interview Definition

> **Hierarchical Clustering is an unsupervised learning technique that creates a hierarchy of clusters by either progressively merging smaller clusters into larger clusters or recursively dividing larger clusters into smaller clusters. The resulting hierarchy is commonly visualized using a dendrogram.**

---

# 2. Types of Hierarchical Clustering

There are two main types:

```text
Hierarchical Clustering
│
├── Agglomerative Clustering
│   └── Bottom-Up Approach
│
└── Divisive Clustering
    └── Top-Down Approach
```

---

# 3. Agglomerative Clustering

This is the **most commonly used** hierarchical clustering approach.

### Working

Initially, every data point is considered an individual cluster.

Then, the two closest clusters are merged repeatedly.

```text
A    B    C    D

Step 1:
A    B    C    D

Step 2:
AB   C    D

Step 3:
AB   CD

Step 4:
ABCD
```

### Process

1. Treat every data point as a separate cluster.
2. Calculate the distance between clusters.
3. Merge the closest two clusters.
4. Repeat until all points belong to one hierarchy.
5. Use a dendrogram to decide the final number of clusters.

---

# 4. Divisive Hierarchical Clustering

Divisive clustering works in the opposite direction.

Initially, all data points belong to one cluster.

The algorithm repeatedly divides clusters.

```text
        ABCD
          │
       ┌──┴──┐
      AB     CD
      ││     ││
      A B    C D
```

### Process

1. Start with one large cluster.
2. Find the best way to split it.
3. Continue splitting clusters.
4. Stop when the desired number of clusters is obtained.

---

# 5. What is a Dendrogram?

A **dendrogram** is a tree-like diagram that shows how clusters are merged.

Example:

```text
Distance
   │
10 │        ─────────────
   │       │
 8 │   ────┘       ──────
   │  │           │
 5 │──┘       ────┘
   │
 0 └──────────────────────
      A   B   C   D
```

The vertical axis represents the **distance or dissimilarity** at which clusters are merged.

### How to Select the Number of Clusters?

Draw a horizontal line through the dendrogram.

The number of vertical branches intersected by that line represents the number of clusters.

A common approach is to choose a cut before a large jump in linkage distance.

# 6. Distance Measures

Hierarchical Clustering can use different distance metrics.

## Euclidean Distance

[
d(x,y)=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
]

## Manhattan Distance

[
d(x,y)=\sum_{i=1}^{n}|x_i-y_i|
]

The choice depends on the dataset and problem.

---

# 7. Linkage Methods

A key concept in Hierarchical Clustering is **linkage**, which determines the distance between clusters.

## 1. Single Linkage

Uses the distance between the **closest points** of two clusters.

[
D(A,B)=\min_{a\in A,b\in B}d(a,b)
]

**Advantage:** Can detect irregular shapes.

**Disadvantage:** Can suffer from the **chaining effect**.

---

## 2. Complete Linkage

Uses the distance between the **farthest points**.

[
D(A,B)=\max_{a\in A,b\in B}d(a,b)
]

Produces more compact clusters.

---

## 3. Average Linkage

Uses the average distance between all pairs of points from two clusters.

[
D(A,B)=\frac{1}{|A||B|}
\sum_{a\in A}\sum_{b\in B}d(a,b)
]

---

## 4. Ward Linkage

Ward's method merges clusters that result in the **smallest increase in within-cluster variance**.

It generally works well for compact, roughly spherical clusters.

# 13. Advantages

* No training labels are required.
* Does not require random centroid initialization.
* Produces a hierarchy of clusters.
* Dendrogram helps visualize cluster relationships.
* The final number of clusters can often be explored by cutting the dendrogram at different levels.

---

# 14. Disadvantages

* Computationally expensive for very large datasets.
* Sensitive to distance metrics and linkage methods.
* Sensitive to feature scaling and sometimes outliers.
* Once clusters are merged in standard agglomerative clustering, they generally cannot be undone.

# 15. Hierarchical Clustering vs K-Means

| Feature                | Hierarchical Clustering    | K-Means                          |
| ---------------------- | -------------------------- | -------------------------------- |
| Type                   | Hierarchical               | Centroid-based                   |
| Structure              | Tree/Dendrogram            | Flat clusters                    |
| Initialization         | No random centroids        | Requires centroid initialization |
| Number of clusters     | Can explore via dendrogram | Must specify K                   |
| Large datasets         | Can be expensive           | Usually faster                   |
| Outliers               | Can be sensitive           | Can be sensitive                 |
| Cluster representation | Hierarchy                  | Centroids                        |
