# Clustering Algorithms
   Clustering is an unsupervised machine learning technique used to discover natural groups or patterns in unlabeled data. The objective is to maximize similarity among data points within the same cluster while minimizing similarity between different clusters. Popular clustering algorithms include K-Means, which partitions data into K groups based on centroid distance; Hierarchical Clustering, which creates a tree-like hierarchy of clusters; DBSCAN, which identifies dense regions and can detect noise; and Gaussian Mixture Models, which provide probabilistic cluster membership. The selection of a clustering algorithm depends on the dataset's size, cluster shape, presence of outliers, and whether the number of clusters is known in advance. Common applications include customer segmentation, anomaly detection, image segmentation, recommendation systems, and document grouping.

#### Clustering algorithms Types
 - Partitioned-based clustering
      - Relatively efficient
      - Eg. k-Means, k-Median, Fuxxy c-Means

 - Hierarchical Clustering
      - Produces trees of clusters
      - Eg.  agglomerative, Divisive

 - Density-based Clustering
     - Produces arbitrary shaped clusters
     - Eg. DBSCAN



# K-Means  Clustering Algorithms
K-Means is an unsupervised, centroid-based clustering algorithm used to partition unlabeled data into K distinct clusters. The algorithm begins by initializing K centroids, typically using the K-Means++ initialization method. Each data point is then assigned to the nearest centroid based on a distance measure, commonly Euclidean distance. After assignment, the centroid of each cluster is recalculated as the mean of the points assigned to that cluster. These assignment and update steps are repeated until the centroids converge or the stopping criteria are reached. The objective of K-Means is to minimize the Within-Cluster Sum of Squares, or WCSS, which measures the squared distance between data points and their respective centroids. The value of K can be selected using techniques such as the Elbow Method, Silhouette Score, and domain knowledge. K-Means is computationally efficient and widely used for applications such as customer segmentation and image segmentation. However, it requires the number of clusters to be chosen in advance, is sensitive to outliers and feature scaling, and generally performs best when clusters are compact and reasonably well separated.

#### How to identify no. of clusters or k value in k-mean clustering?
   In K-Means clustering, the optimal number of clusters, K, is commonly determined using the Elbow Method and Silhouette Score. In the Elbow Method, I train the K-Means algorithm for different values of K and plot K against WCSS or inertia. The optimal K is typically selected at the point where the reduction in WCSS starts to slow significantly, forming an elbow. I also validate the result using the Silhouette Score, where a higher score indicates better cohesion within clusters and better separation between clusters. In practice, I combine these quantitative methods with domain knowledge to select a K value that is both statistically meaningful and useful for the business problem.

# 3. How K-Means Works

```text
Unlabeled Dataset
       ↓
Choose Number of Clusters (K)
       ↓
Initialize K Centroids
       ↓
Calculate Distance
       ↓
Assign Points to Nearest Centroid
       ↓
Recalculate Centroids
       ↓
Repeat Until Convergence
```

---

# 4. Step-by-Step Working

Assume:

```text
K = 2
```

## Step 1: Initialize Centroids

The algorithm selects **K initial centroids**.

```text
Data Points:

● ● ● ● ● ●

Centroid 1 = C1
Centroid 2 = C2
```

---

## Step 2: Calculate Distance

For every data point, calculate its distance from each centroid.

The most commonly used distance is **Euclidean Distance**.

### Formula

[
d(x,c)=\sqrt{\sum_{i=1}^{n}(x_i-c_i)^2}
]

For two features:

[
d=\sqrt{(x_1-c_1)^2+(x_2-c_2)^2}
]

The point is assigned to the **nearest centroid**.

---

## Step 3: Create Clusters

```text
Cluster 1:

● ● ●

Cluster 2:

● ● ●
```

---

## Step 4: Update Centroids

The new centroid is the **mean of all points in that cluster**.

[
\mu_j=\frac{1}{n_j}\sum_{i=1}^{n_j}x_i
]

Where:

* (\mu_j) = centroid of cluster (j)
* (n_j) = number of points in cluster (j)

---

## Step 5: Repeat

The algorithm repeats:

1. Assign points to nearest centroid.
2. Calculate new centroids.

Until the centroids stop changing significantly.

This is called **convergence**.

---

# 5. Objective Function – WCSS / Inertia

K-Means tries to minimize the **Within-Cluster Sum of Squares (WCSS)**.

[
WCSS=
\sum_{j=1}^{K}
\sum_{x_i \in C_j}
||x_i-\mu_j||^2
]

Where:

* (K) = number of clusters
* (C_j) = cluster (j)
* (x_i) = data point
* (\mu_j) = centroid of the cluster

### Meaning

K-Means tries to ensure that data points within the same cluster are as close as possible to their centroid.

In scikit-learn:

```python
kmeans.inertia_
```

returns the WCSS value.

# 6. How to Choose the Best Value of K?

## Elbow Method

The Elbow Method calculates WCSS for different values of K.

```text
K = 1 → High WCSS
K = 2 → Lower WCSS
K = 3 → Lower WCSS
K = 4 → Significant decrease
K = 5 → Small decrease

                    ← Elbow Point
```

The K value at which the improvement starts decreasing significantly is often selected.

# Frequently Asked Interview Questions

### Q1. What is K-Means Clustering?

> **K-Means is an unsupervised learning algorithm that partitions unlabeled data into K clusters by assigning each data point to the nearest centroid and iteratively updating the centroids until convergence.**

---

### Q2. Why is K-Means an Unsupervised Algorithm?

> **K-Means works on unlabeled data. There is no target variable or predefined output class; the algorithm discovers groups based on similarity.**

---

### Q3. What does K represent?

> **K represents the number of clusters that we want the algorithm to create.**

---

### Q4. What is a centroid?

> **A centroid is the mean position of all data points belonging to a particular cluster and represents the center of that cluster.**

---

### Q5. What is WCSS?

> **WCSS, or Within-Cluster Sum of Squares, measures the total squared distance between each data point and its assigned cluster centroid. K-Means minimizes WCSS to create compact clusters.**

---

### Q6. How do you select K?

> **Common methods include the Elbow Method, Silhouette Score, and domain knowledge. The Elbow Method looks for the point where increasing K produces diminishing improvements in WCSS.**

---

### Q7. Does K-Means require feature scaling?

> **Feature scaling is generally recommended because K-Means relies on distance calculations, and features with larger numerical ranges can dominate the clustering process.**

---

### Q8. What is the difference between `fit()` and `fit_predict()`?

Trains the model.

```python
labels = kmeans.fit_predict(X)
```

Trains the model and directly returns the cluster labels.
