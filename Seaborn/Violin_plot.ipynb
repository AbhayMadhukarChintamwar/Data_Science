```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df  = sns.load_dataset('tips')
df
```

```python
sns.violinplot(x='time' , y = 'total_bill' , data = df)
plt.show()
```

```python
sns.violinplot(x='time' , y = 'total_bill' , data = df, hue='sex' , inner = 'quartz')
plt.show()
```

```python
sns.violinplot(x='time' , y = 'total_bill' , data = df , order =['Dinner', 'Lunch'], inner = 'quartz', palette='rainbow_r')
plt.show()
```

```python
sns.violinplot(x='time' , y = 'total_bill' , data = df , order =['Dinner', 'Lunch'], inner = 'box', palette='rainbow_r')
plt.show()
```

```python
sns.violinplot(x='time' , y = 'total_bill' , data = df , order =['Dinner', 'Lunch'], inner = 'point', palette='rainbow_r')
plt.show()
```

```python
sns.violinplot(x='time' , y = 'total_bill' , data = df , order =['Dinner', 'Lunch'], inner = 'stick', palette='rainbow_r')
plt.show()
```
