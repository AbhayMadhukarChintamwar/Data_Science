CAT PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, palette='rainbow_r')
plt.show()
```

```python
df = sns.load_dataset('tips').head(20)
df
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, palette='rainbow_r')
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex')
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex', palette='hot_r')
plt.grid()
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex', palette='hot_r')
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex', palette='seismic', height=3)
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex', palette='seismic', height=10)
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex', palette='seismic', height=10, kind='bar')
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'time', y = 'total_bill', data = df, hue = 'sex', palette='seismic', height=10, kind='box')
plt.xticks(rotation = 'vertical')
plt.show()
```

```python
sns.catplot(x = 'tip', y = 'size', data = df, hue = 'sex', palette='seismic', height=10, kind='point')
plt.xticks(rotation = 'vertical')
plt.show()
```
