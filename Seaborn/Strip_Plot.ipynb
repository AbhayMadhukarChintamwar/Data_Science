STRIP PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.stripplot(df)
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df)
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df, palette='rainbow_r')
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r')
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex')
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r', linewidth=1)
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r', linewidth=1,  edgecolor='r')
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r', linewidth=1,  edgecolor='r', jitter=2)
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r', linewidth=1,  edgecolor='r', size = 10)
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r', linewidth=1,  edgecolor='g', marker ='<')
plt.show()
```

```python
sns.stripplot(x= 'day', y ='total_bill', data=df,  hue = 'sex', palette='rainbow_r', linewidth=1,  edgecolor='g', marker ='<' , alpha = 0.5)
plt.show()
```

```python
sns.stripplot(x = df['total_bill'])
plt.show()
```

```python
sns.stripplot(y = df['total_bill'])
plt.show()
```
