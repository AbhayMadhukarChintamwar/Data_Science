BOX PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.boxplot(x = 'day', y ='total_bill', data= df)
plt.show()
```

```python
sns.boxplot(x = 'day', y ='total_bill', data= df, hue = 'sex')
plt.show()
```

```python
sns.boxplot(x = 'day', y ='total_bill', data= df, palette='rainbow_r' )
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(x = 'day', y ='total_bill', data= df, palette='rainbow_r' )
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(x = 'day', y ='total_bill', data= df, hue = 'time', palette='rainbow_r' )
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(x = 'day', y ='total_bill', data= df, hue = 'time' )
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(x = 'day', y ='total_bill', data= df, hue = 'sex', order=['Fri', 'Sun','Thur', ] )
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(x = 'day', y ='total_bill', data= df, hue = 'sex', order=['Fri', 'Sun','Thur', ] , showmeans = True )
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(x = 'day', y ='total_bill', data= df, hue = 'sex' , showmeans = True,
            meanprops ={'marker':'+', 'markeredgecolor': 'blue'} , palette='rainbow_r')
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(data= df,  showmeans = True,
            meanprops ={'marker':'+', 'markeredgecolor': 'blue'} , palette='rainbow_r')
plt.show()
```

```python
sns.set(style='whitegrid')
sns.boxplot(data= df,  showmeans = True,
            meanprops ={'marker':'+', 'markeredgecolor': 'blue'} , palette='rainbow_r', orient='h')
plt.show()
```

```python
sns.boxplot(y = df['total_bill'])
plt.show()
```
