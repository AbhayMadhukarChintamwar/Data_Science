PAIR PLOT

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = sns.load_dataset('tips')
df
```

```python
sns.pairplot(df)
plt.show()
```

```python
sns.pairplot(df, hue='sex')
plt.show()
```

```python
sns.pairplot(df, vars =['total_bill', 'tip'], hue='sex')
plt.show()
```

```python
sns.pairplot(df, hue='sex', hue_order=["Female", "Male"])
plt.show()
```

```python
sns.pairplot(df, hue='sex', palette='rainbow_r')
plt.show()
```

```python
sns.pairplot(df, hue='sex', hue_order=["Female", "Male"], palette='rainbow_r')
plt.show()
```

```python
sns.pairplot(df, vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='rainbow_r')
plt.show()
```

```python
sns.pairplot(df, x_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='rainbow_r')
plt.show()
```

```python
sns.pairplot(df, y_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='rainbow_r')
plt.show()
```

```python
sns.pairplot(df, y_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='rainbow_r',
             kind='reg')
plt.show()
```

```python
sns.pairplot(df, y_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='rainbow_r',
             kind='hist')
plt.show()
```

```python
sns.pairplot(df, y_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='gist_rainbow_r',
             kind='hist')
plt.show()
```

```python
sns.pairplot(df, y_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='gist_rainbow_r',
             kind='kde')
plt.show()
```

```python
sns.pairplot(df, y_vars =['total_bill', 'tip'], hue='sex', hue_order=["Female", "Male"], palette='gist_rainbow_r',
             kind='kde', diag_kind='hist')
plt.show()
```

```python
sns.pairplot(df, hue='sex', hue_order=["Female", "Male"], palette='gist_rainbow_r',
             kind='kde', diag_kind='hist')
plt.show()
```

```python
sns.pairplot(df, hue='sex', hue_order=["Female", "Male"], palette='gist_rainbow_r',
             kind='kde', diag_kind='hist')
plt.show()
```
