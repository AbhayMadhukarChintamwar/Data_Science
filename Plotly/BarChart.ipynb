BAR CHART

```python
import plotly.graph_objs as go
```

```python
label = ['Cricket','Football','Hockey','Basketball']
revenue = [400,200,200,300]
trace = go.Bar(x = label, y=revenue)
go.Figure(trace)
```

```python
label = ['Cricket','Football','Hockey','Basketball']
revenue = [400,200,200,300]
trace1 = go.Bar(x = label, y=revenue)
trace2=go.Bar(x=label, y = [300,200,350,250])
go.Figure(data =[trace1, trace2])
```

```python
label = ['Cricket','Football','Hockey','Basketball']
revenue = [400,200,200,300]
trace1 = go.Bar(x = label, y=revenue, name='2025')
trace2=go.Bar(x=label, y = [300,200,350,250],name='2026')
go.Figure(data =[trace1, trace2])
```

```python
label = ['Cricket','Football','Hockey','Basketball']
revenue = [400,200,200,300]
trace1 = go.Bar(x = label, y=revenue, name='2025')
trace2=go.Bar(x=label, y = [300,200,350,250],name='2026')
layout = go.Layout(title='Sport Revenue by Year', xaxis_title = 'Sports', yaxis_title = 'Revenue')
go.Figure(data =[trace1, trace2], layout=layout)
```
